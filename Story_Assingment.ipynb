```python
year_row   = raw.iloc[1].ffill()
area_row   = raw.iloc[2].ffill()
gender_row = raw.iloc[3]
data       = raw.iloc[4:42]

rows_out = []
for i, row in data.iterrows():
    state = row[0]
    if pd.isna(state):
        continue
    for col in range(1, len(row)):
        val = row[col]
        if pd.isna(val):
            continue
        rows_out.append({
            'state': state,
            'year': str(year_row[col]),
            'area': area_row[col],
            'gender': gender_row[col],
            'value': val
        })

df = pd.DataFrame(rows_out)
df.to_csv(r"C:\Users\sanvi\Downloads\data acj\sdg9_manufacturing_employment_tidy.csv", index=False)
df.head()
```

```python
latest_year = df['year'].unique()[-1]
top_states = (df[(df['year'] == latest_year) & (df['area'] == 'rural + urban') & (df['gender'] == 'person')]
              .sort_values('value', ascending=False)
              .head(10))
top_states[['state', 'value']].reset_index(drop=True)
```

```python
gender_gap = (df[(df['year'] == latest_year) & (df['area'] == 'rural + urban') & (df['gender'].isin(['male','female'])) & (df['state'] != 'India')]
              .pivot(index='state', columns='gender', values='value'))
gender_gap['gap_female_minus_male'] = gender_gap['female'] - gender_gap['male']
gender_gap.sort_values('gap_female_minus_male', ascending=False).head(10)
```

```python
state_data = df[(df['year'] == latest_year) & (df['area'] == 'rural + urban') & (df['gender'] == 'person') & (df['state'] != 'India')]

highest = state_data.loc[state_data['value'].idxmax()]
lowest = state_data.loc[state_data['value'].idxmin()]

print(f"Highest manufacturing employment share: {highest['state']} at {highest['value']}% ({latest_year})")
print(f"Lowest manufacturing employment share: {lowest['state']} at {lowest['value']}% ({latest_year})")
```

```python
up_data = df[(df['state'] == 'Uttar Pradesh') & (df['area'] == 'rural + urban') & (df['gender'] == 'person')]
up_data = up_data.sort_values('year')
up_data[['year', 'value']].reset_index(drop=True)
```

```python
up_rank = state_data.sort_values('value', ascending=False).reset_index(drop=True)
up_position = up_rank[up_rank['state'] == 'Uttar Pradesh'].index[0] + 1
up_value = up_rank[up_rank['state'] == 'Uttar Pradesh']['value'].values[0]

print(f"Uttar Pradesh ranks #{up_position} out of {len(up_rank)} states/UTs with a manufacturing employment share of {up_value}% in {latest_year}.")
```

```python
union_territories = ['A & N Islands', 'Chandigarh', 'Dadra & Nagar Haveli', 'Daman & Diu',
                      'Delhi', 'Jammu & Kashmir', 'Ladakh', 'Lakshadweep', 'Puducherry']

latest_data = df[(df['year'] == latest_year) & (df['area'] == 'rural + urban') & (df['gender'] == 'person') & (df['state'] != 'India')]

states_table = latest_data[~latest_data['state'].isin(union_territories)].sort_values('value', ascending=False)
uts_table = latest_data[latest_data['state'].isin(union_territories)].sort_values('value', ascending=False)

print("STATES — Manufacturing Employment Share")
display(states_table[['state', 'value']].reset_index(drop=True))

print("\nUNION TERRITORIES — Manufacturing Employment Share")
display(uts_table[['state', 'value']].reset_index(drop=True))
```

```python
print(f"\nAverage — States: {states_table['value'].mean():.2f}%")
print(f"Average — UTs: {uts_table['value'].mean():.2f}%")
```
