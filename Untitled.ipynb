```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
file_path = r"C:\Users\sanvi\Downloads\data acj\sdg9_manufacturing_employment_tidy.csv"

df = pd.read_csv(file_path)

print(df.shape)
print(df.head())
print(df.columns)
```

```python
df["value"] = pd.to_numeric(df["value"], errors="coerce")

# Remove rows where value is missing
df = df.dropna(subset=["value"])

# Clean text columns
df["state"] = df["state"].astype(str).str.strip()
df["area"] = df["area"].astype(str).str.strip().str.lower()
df["gender"] = df["gender"].astype(str).str.strip().str.lower()
df["year"] = df["year"].astype(str).str.strip()

print("\nCleaned dataset shape:", df.shape)

print("\nYears available:")
print(df["year"].unique())
```

```python
person_df = df[
    (df["gender"] == "person") &
    (df["area"] == "rural + urban")
].copy()

print("\nPerson + Rural & Urban dataset:")
print(person_df.head())
```

```python
# ------------------------------------------------------------
# 1. MAIN DATASET
# ------------------------------------------------------------

person_df = df[
    (df["gender"] == "person") &
    (df["area"] == "rural + urban")
].copy()

# Get available years
years = person_df["year"].unique()

year_order = [
    "2017-18",
    "2018-19",
    "2019-20",
    "2020-21",
    "2021-22",
    "2022-23",
    "2023-24",
    "2025"
]

available_years = [y for y in year_order if y in years]

latest_year = available_years[-1]
first_year = available_years[0]

print("Available years:", available_years)
print("First year:", first_year)
print("Latest year:", latest_year)
print("Number of states/UTs:", person_df["state"].nunique())
```

```python
# ------------------------------------------------------------
# 2. LATEST YEAR: STATE/UT COMPARISON
# ------------------------------------------------------------

latest = person_df[
    person_df["year"] == latest_year
].copy()

latest = latest.sort_values(
    "value",
    ascending=False
)

print(
    latest[
        ["state", "value"]
    ].to_string(index=False)
)
```

```python
# ------------------------------------------------------------
# 3. TOP AND BOTTOM STATES/UTs
# ------------------------------------------------------------

print("TOP 10")
print("-" * 40)

print(
    latest.head(10)[
        ["state", "value"]
    ].to_string(index=False)
)

print("\nBOTTOM 10")
print("-" * 40)

print(
    latest.tail(10).sort_values("value")[
        ["state", "value"]
    ].to_string(index=False)
)
```

```python
# ------------------------------------------------------------
# 4. HOW WIDE IS THE GAP?
# ------------------------------------------------------------

highest = latest.iloc[0]
lowest = latest.iloc[-1]

gap = highest["value"] - lowest["value"]

print("Highest:", highest["state"], highest["value"], "%")
print("Lowest:", lowest["state"], lowest["value"], "%")
print("Gap:", round(gap, 2), "percentage points")

print(
    "Average across states/UTs:",
    round(latest["value"].mean(), 2),
    "%"
)

print(
    "Median:",
    round(latest["value"].median(), 2),
    "%"
)
```

```python
# ------------------------------------------------------------
# 5. CHANGE OVER TIME
# ------------------------------------------------------------

first = person_df[
    person_df["year"] == first_year
][["state", "value"]].copy()

first = first.rename(
    columns={"value": "first_value"}
)

last = person_df[
    person_df["year"] == latest_year
][["state", "value"]].copy()

last = last.rename(
    columns={"value": "latest_value"}
)

change = pd.merge(
    first,
    last,
    on="state",
    how="inner"
)

change["change_percentage_points"] = (
    change["latest_value"] -
    change["first_value"]
)

change = change.sort_values(
    "change_percentage_points",
    ascending=False
)

print(
    change[
        [
            "state",
            "first_value",
            "latest_value",
            "change_percentage_points"
        ]
    ].to_string(index=False)
)
```

```python
# ------------------------------------------------------------
# 6. BIGGEST CHANGES
# ------------------------------------------------------------

print("BIGGEST INCREASES")
print("-" * 50)

print(
    change.head(10)[
        [
            "state",
            "first_value",
            "latest_value",
            "change_percentage_points"
        ]
    ].to_string(index=False)
)

print("\nBIGGEST DECLINES")
print("-" * 50)

print(
    change.tail(10).sort_values(
        "change_percentage_points"
    )[
        [
            "state",
            "first_value",
            "latest_value",
            "change_percentage_points"
        ]
    ].to_string(index=False)
)
```

```python
# ------------------------------------------------------------
# 7. RURAL VS URBAN
# ------------------------------------------------------------

rural_urban = df[
    (df["gender"] == "person") &
    (df["year"] == latest_year) &
    (df["area"].isin(["rural", "urban"]))
].copy()

ru_pivot = rural_urban.pivot_table(
    index="state",
    columns="area",
    values="value",
    aggfunc="mean"
)

ru_pivot["urban_minus_rural"] = (
    ru_pivot["urban"] -
    ru_pivot["rural"]
)

print(
    ru_pivot.sort_values(
        "urban_minus_rural",
        ascending=False
    ).round(2).to_string()
)
```

```python
# ------------------------------------------------------------
# GRAPH 1: TOP 15 STATES/UTs
# ------------------------------------------------------------

top15 = latest.head(15).sort_values("value")

plt.figure(figsize=(10, 7))

plt.barh(
    top15["state"],
    top15["value"]
)

plt.xlabel(
    "Manufacturing employment as share of total employment (%)"
)

plt.ylabel("State / Union Territory")

plt.title(
    f"Manufacturing's share of employment varies across India ({latest_year})"
)

plt.tight_layout()

plt.savefig(
    "graph1_top15_manufacturing_share.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ------------------------------------------------------------
# GRAPH 2: ALL STATES/UTs
# ------------------------------------------------------------

all_states = latest.sort_values("value")

plt.figure(figsize=(11, 14))

plt.barh(
    all_states["state"],
    all_states["value"]
)

plt.xlabel(
    "Manufacturing employment as share of total employment (%)"
)

plt.ylabel("State / Union Territory")

plt.title(
    f"Manufacturing employment share across Indian states and UTs ({latest_year})"
)

plt.tight_layout()

plt.savefig(
    "graph2_all_states_manufacturing_share.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ------------------------------------------------------------
# GRAPH 3: TREND
# ------------------------------------------------------------

trend = (
    person_df
    .groupby("year")["value"]
    .mean()
    .reindex(available_years)
    .reset_index()
)

trend.columns = [
    "year",
    "average_share"
]

plt.figure(figsize=(10, 6))

plt.plot(
    trend["year"],
    trend["average_share"],
    marker="o",
    linewidth=2
)

plt.xlabel("Year")

plt.ylabel(
    "Average manufacturing employment share (%)"
)

plt.title(
    "Average manufacturing employment share across states and UTs"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    "graph3_manufacturing_share_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ------------------------------------------------------------
# GRAPH 4: CHANGE OVER TIME
# ------------------------------------------------------------

change_graph = change.sort_values(
    "change_percentage_points"
)

plt.figure(figsize=(11, 14))

plt.barh(
    change_graph["state"],
    change_graph["change_percentage_points"]
)

plt.axvline(
    0,
    linewidth=1
)

plt.xlabel(
    f"Change in manufacturing employment share ({first_year} to {latest_year}) — percentage points"
)

plt.ylabel("State / Union Territory")

plt.title(
    "Change in manufacturing's share of employment across India"
)

plt.tight_layout()

plt.savefig(
    "graph4_change_manufacturing_share.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ------------------------------------------------------------
# GRAPH 5: RURAL VS URBAN
# ------------------------------------------------------------

ru_graph = ru_pivot.sort_values(
    "urban_minus_rural",
    ascending=False
).head(15)

x = np.arange(len(ru_graph))
width = 0.35

plt.figure(figsize=(12, 7))

plt.bar(
    x - width / 2,
    ru_graph["rural"],
    width,
    label="Rural"
)

plt.bar(
    x + width / 2,
    ru_graph["urban"],
    width,
    label="Urban"
)

plt.xticks(
    x,
    ru_graph.index,
    rotation=60,
    ha="right"
)

plt.ylabel(
    "Manufacturing employment as share of total employment (%)"
)

plt.xlabel("State / Union Territory")

plt.title(
    f"Manufacturing's employment share: rural vs urban ({latest_year})"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    "graph5_rural_vs_urban.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
